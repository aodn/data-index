```python
# --- Local SQLite Iceberg config ---
local_warehouse = pathlib.Path(".load/orchestrate-local")
data_index_catalog_config = SqliteCatalogConfig(
    name="data-index",
    uri=f"sqlite:///{(local_warehouse / 'catalog.db').resolve()}",
    warehouse=str(local_warehouse.resolve()),
)

structured_table = IcebergTableConfig(
    catalog_config=data_index_catalog_config,
    namespace="data_index",
    table_name="structured_metadata_v5",
).load()

unstructured_table = IcebergTableConfig(
    catalog_config=data_index_catalog_config,
    namespace="data_index",
    table_name="unstructured_metadata_v4",
).load()
```

```python
# Read into memory as polars.DataFrame
sdf = structured_table.scan().to_polars()
udf = unstructured_table.scan().to_polars()
display(sdf)
display(udf)
```
