```python
import pathlib

import polars
import rich

from data_index.iceberg_config import SqliteCatalogConfig, IcebergTableConfig
```

```python
# --- Local SQLite Iceberg config ---
local_warehouse = pathlib.Path("/Users/thommodin/dev/data-index/.load/orchestrate-local")
local_warehouse.mkdir(parents=True, exist_ok=True)

data_index_catalog_config = SqliteCatalogConfig(
    name="data-index",
    uri=f"sqlite:///{local_warehouse / 'catalog.db'}",
    warehouse=str(local_warehouse),
)

table = IcebergTableConfig(
    catalog_config=data_index_catalog_config,
    namespace="data_index",
    table_name="structured_metadata_v6",
).load()
```

```python
# Read into memory as polars.DataFrame
df = table.scan().to_polars()
display(df)
```

```python
display(df["coordinate_schema"].value_counts(sort=True))
display(df["variable_schema"].value_counts(sort=True))
display(df["schema_version"].value_counts(sort=True))
```

```python
# Group by variable_schema to get distinct schemas and their row counts (numerosity)
schema_counts = (
    df.select("variable_schema")
    .drop_nulls()
    .group_by("variable_schema")
    .len(name="schema_count")
)

unique_schemas = schema_counts["variable_schema"].to_list()
counts = schema_counts["schema_count"].to_list()

# Extract Core Schema (tuples appearing across ALL distinct schemas)
num_unique_schemas = len(unique_schemas)

core_tuples = set.intersection(*(
    {(var["key"], var["value"]) for var in schema}
    for schema in unique_schemas
)) if unique_schemas else set()

core_schema = [{"key": k, "value": v} for k, v in core_tuples]

# Build Non-Core per schema, attaching the schema's numerosity (count)
non_core_per_schema = [
    {
        "schema_count": count,
        "non_core_elements": [
            var for var in schema 
            if (var["key"], var["value"]) not in core_tuples
        ]
    }
    for schema, count in zip(unique_schemas, counts)
]
rich.print(core_schema)
rich.print(non_core_per_schema)
```
